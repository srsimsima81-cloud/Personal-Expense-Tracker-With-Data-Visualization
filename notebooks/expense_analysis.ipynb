## 📊 Dataset Info

## 💰 Category-wise Spending

```python
cat = category_analysis(df)
cat
```

## 📅 Monthly Spending Trend

```python
month = monthly_analysis(df)
month
```

```python
month.plot(marker='o', color='green')
plt.title('Monthly Trend')
plt.show()
```

## 💳 Payment Method Distribution

```python
pay = payment_analysis(df)
pay
```

```python
pay.plot(kind='pie', autopct='%1.1f%%')
plt.title('Payment Methods')
plt.show()
```

```python
print('Highest spending category:', cat.idxmax())
print('Total spending:', df['Amount'].sum())
print('Average daily spending:', df.groupby('Date')['Amount'].sum().mean())
```
